```python
import random
import pyautogui as pg
import time

# List of animals
animals = ('madda gudu', 'lanja', 'puku', 'koduku')

# Wait 8 seconds to give time to switch to the target window
time.sleep(10)

# Number of messages to send
messages_to_send = 500 # Reduce from 500 for safer testing

for i in range(messages_to_send):
    try:
        animal = random.choice(animals)
        pg.write(f"You are {animal}")
        pg.press('enter')
        time.sleep(1)  # Add delay between messages
    except Exception as e:
        print(f"An error occurred: {e}")
        break
```
